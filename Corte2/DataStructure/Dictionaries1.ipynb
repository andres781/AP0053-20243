```python
#Python permite crear un diccionario usando la siguiente sintaxis
names = ['Jenyy', 'Alexus', 'Sam', 'Grace']
heights = [61, 70, 67, 64]

## zip() combina dos listas en un iterador de tuplas con elementos de lista emparejados. Este dictado de comprensión:

zipStudents = zip(names, heights)#Diccionario zipStudents
print("zipStudents: ", zipStudents)

students = {key:value for key, value in zip(names, heights)}
##students is now {'Jenny': 61, 'Alexus': 70, 'Sam':67, 'Grace':64}
print(students)

drinks = ["espresso", "chai", "decaf", "drips"]
caffeine = [64, 40, 0, 120]

zipped_drinks = zip(drinks, caffeine)
print(zipped_drinks)

drinks_to_caffeine = {key:value for key, value in zipped_drinks}
print(drinks_to_caffeine)
```
